```python
from google.colab import files
import pandas as pd

uploaded = files.upload()  # Upload emoji_df.csv when prompted
df = pd.read_csv("emoji_df.csv")
print(f"Loaded {len(df)} emojis!")
print(df.head())
```

```python
from transformers import pipeline

emotion_model = pipeline(
    "text-classification",
    model="SamLowe/roberta-base-go_emotions",
    top_k=3
)

print("✅ Model loaded!")
```

```python
# Groups similar emotions together so scoring is fair
emotion_groups = {
    "joy":      ["joy", "amusement", "excitement", "optimism", "approval",
                 "pride", "relief", "gratitude", "admiration", "caring", "love"],
    "sadness":  ["sadness", "grief", "disappointment", "remorse"],
    "anger":    ["anger", "annoyance", "disapproval"],
    "fear":     ["fear", "nervousness"],
    "surprise": ["surprise", "realization", "confusion", "curiosity"],
    "disgust":  ["disgust"],
    "neutral":  ["neutral"],
}

# Flip it so we can look up any emotion and find its group
emotion_to_group = {}
for group, members in emotion_groups.items():
    for emotion in members:
        emotion_to_group[emotion] = group

# Map each group to emoji sub-groups from your CSV
group_to_subgroup = {
    "joy":      ["face-smiling", "face-affection", "heart"],
    "sadness":  ["face-concerned", "face-sleepy"],
    "anger":    ["face-negative", "face-hand"],
    "fear":     ["face-concerned", "face-unwell"],
    "surprise": ["face-hand", "face-neutral-skeptical"],
    "disgust":  ["face-negative", "face-unwell"],
    "neutral":  ["face-neutral-skeptical"],
}

print("✅ Emotion groups and emoji map ready!")
```

```python
import random

def predict_emoji(text):
    # Step 1: Get top 3 emotions from the model
    results = emotion_model(text)[0]

    # Step 2: Try each emotion until we find a matching emoji
    chosen_emoji = None
    chosen_name = None
    chosen_emotion = None
    chosen_group = None
    chosen_confidence = None

    for result in results:
        raw_emotion = result["label"]
        confidence = result["score"]
        group = emotion_to_group.get(raw_emotion, "neutral")
        target_subgroups = group_to_subgroup.get(group, ["face-neutral-skeptical"])
        matches = df[df["sub_group"].isin(target_subgroups)]

        if len(matches) > 0:
            chosen = matches.sample(1).iloc[0]
            chosen_emoji = chosen["emoji"]
            chosen_name = chosen["name"]
            chosen_emotion = raw_emotion
            chosen_group = group
            chosen_confidence = confidence
            break

    if chosen_emoji is None:
        chosen_emoji, chosen_name, chosen_emotion, chosen_group, chosen_confidence = "🤔", "thinking face", "unknown", "unknown", 0.0

    print(f"📝 Input:    '{text}'")
    print(f"🔍 Detected: {chosen_emotion} → grouped as '{chosen_group}' ({chosen_confidence:.0%} confident)")
    print(f"✅ Emoji:    {chosen_emoji}  ({chosen_name})")
    print("-" * 45)

print("✅ Prediction function ready!")
```

```python
predict_emoji("I just got an A on my test!")
predict_emoji("I miss my dog so much.")
predict_emoji("There's a spider on my arm!!")
predict_emoji("I don't really care either way.")
predict_emoji("I can't believe they threw me a surprise party!")
predict_emoji("I am so proud of what I made!")
predict_emoji("Ugh this is so annoying.")
predict_emoji("I feel so loved by my friends.")
predict_emoji("That was so gross I almost threw up.")
```

```python
test_cases = [
    # Joy
    ("I am so happy today!", "joy"),
    ("This made me smile so big!", "joy"),
    ("I love spending time with my friends!", "joy"),
    ("Best day of my life!", "joy"),
    ("I'm so excited for the weekend!", "joy"),
    # Sadness
    ("I am devastated right now.", "sadness"),
    ("I cried watching that movie.", "sadness"),
    ("This is the worst day ever.", "sadness"),
    ("I feel so alone.", "sadness"),
    ("I really miss my grandma.", "sadness"),
    # Anger
    ("I am furious at what happened.", "anger"),
    ("That really ticked me off.", "anger"),
    ("I can't stand this anymore.", "anger"),
    ("This is so unfair and I hate it.", "anger"),
    # Fear
    ("That noise scared me so much.", "fear"),
    ("I'm really nervous about tomorrow.", "fear"),
    ("I'm terrified of failing.", "fear"),
    # Surprise
    ("Wow, I did not expect that at all!", "surprise"),
    ("No way, are you serious?!", "surprise"),
    ("I have no idea what just happened.", "surprise"),
    # Disgust
    ("That smells absolutely horrible.", "disgust"),
    ("I can't even look at that.", "disgust"),
    ("That is so gross I want to leave.", "disgust"),
    # Neutral
    ("I went to the store today.", "neutral"),
    ("The meeting is at 3pm.", "neutral"),
    ("I have math class tomorrow.", "neutral"),
]

correct = 0
wrong_cases = []

print("🧪 Evaluation Results:")
print("=" * 55)

for text, expected in test_cases:
    top_results = emotion_model(text)[0]
    predicted_raw = top_results[0]["label"]
    predicted_group = emotion_to_group.get(predicted_raw, predicted_raw)
    is_correct = predicted_group == expected

    if is_correct:
        correct += 1
    else:
        wrong_cases.append((text, expected, predicted_raw, predicted_group))

    status = "✅" if is_correct else "❌"
    print(f"{status} '{text}'")
    print(f"   Expected: {expected} | Got: {predicted_raw} (counts as: {predicted_group})")

print("=" * 55)
print(f"🎯 Accuracy: {correct}/{len(test_cases)} = {correct/len(test_cases):.0%}")

if wrong_cases:
    print("\n📋 Sentences that got it wrong:")
    for text, expected, raw, group in wrong_cases:
        print(f"  ❌ '{text}'")
        print(f"     Expected '{expected}' but got '{raw}' (counted as '{group}')")
```

```python
while True:
    text = input("\nType a sentence (or 'quit' to stop): ")
    if text.lower() == "quit":
        print("👋 Bye!")
        break
    predict_emoji(text)
```
